```python
import pandas as pd
from utils import Helper
from pathlib import Path
import os, re
utils = Helper()
```

```python
rotate_json = r"C:\Users\kaustubh.keny\Projects\OFFICE PROJECTS\eda_pdf\v2.json"
json_dir = r"C:\Q1_2026_RESULTS\v1_json_working"

r_data = utils.load_json(rotate_json)

for pdf_name in r_data.keys():

    j_path = os.path.join(
        json_dir,
        pdf_name.replace(".pdf", ".json")
    )

    data = utils.load_json(j_path)

    # rebuild table_pages from corrected pages
    table_pages = {}

    for record in data["pages"]:

        pgn = str(record["page_number"])

        tables = [
            {
                "page_type": record["page_type"],
                "coordinates": "||".join(
                    box.split("--")[1].split("|")
                ),
                "financial_term_count": 0,
                "numeric_count": 20
            }
            for box in record["layout_boxes"]
            if box.startswith("table--")
        ]

        if tables:
            table_pages[pgn] = tables

    data["table_pages"] = table_pages
    data["table_page_count"] = len(table_pages)
    data["to_parser"] = sorted(map(int, table_pages.keys()))

    utils.save_json(data, j_path)
```
